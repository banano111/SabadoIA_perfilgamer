Crear 2 funciones:

1.   Primer funcion se va a llamar Saludar - Pide tu nombre y te muestra un saludo
2.   Una segunda funcion que calcule tu nivel de experiencia. (Total de XP)





```python
name = input("Cual es tu nombre?: ")
xp = int(input("Cuantos XP tienes?: "))

def saludar(name):
  print(f"Hola, {name}. Espero tengas un gran dia!!")

def calcularLv(xp):
  print("Calculando tu nivel de experiencia....")

  if xp < 100:
    print("Nivel 1 - Principiante")
  elif xp < 200:
    print("Nivel 2 - Intermedio")
  elif xp > 300:
    print("Nivel 3 - Avanzado")
  else:
    print("Este XP no es valido")

saludar(name)
calcularLv(xp)
```
